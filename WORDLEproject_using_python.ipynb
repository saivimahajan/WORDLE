```python
import random
#bug fixed
#Word list
words = ["Apple", "Brick", "Chalk", "Drape", "Elbow", "Flame", "Grape", "Hazel", "Igloo", "Joker", "Knife", "Lemon", "Magic", "Nurse",
         "Oasis", "Plant", "Quirk", "Raven", "Shelf", "Tiger", "Uncle", "Voice", "Wheat", "Xylox", "Youth", "Zebra", "Alley", "Brave",
         "Chase", "Drown", "Ember", "Frost", "Glory", "Hatch", "Ivory", "Jolly", "Knack", "Ledge", "Right", "North", "Opera", "Peace",
         "Quiet", "Royal", "Stone", "Thumb", "Under", "Vivid", "Whirl", "Zesty", "Aroma", "Burst", "Climb", "Drift", "Earth", "Fable",
         "Ghost", "Honey", "Inlet", "Jumpy", "Kneel", "Light", "Merry", "Night", "Ozone", "Pouch", "Quest", "Straw", "Slide", "Trend",
         "Unity", "Visit", "Weave", "Saivi", "Yield", "Zebra", "Adorn", "Blend", "Craft", "Dwell", "Eager", "Fling", "Glaze", "Hinge",
         "Inner", "Juice", "Knead", "Looms", "Match", "Notch", "Orbit", "Prawn", "Quota", "Ridge", "Slate", "Trace", "Urban", "Vigor",
         "Woven", "Wrong", "Young", "Xerox"]

#random word
w = random.choice(words).lower()

print("Welcome to Wordle! \nGuess the 5-letter word in 6 attempts.")
print("A few things to note -")
print("Green: correct letter and correct position.")
print("Yellow: correct letter but in the wrong position.")
print("Black: letter not in the word.")
print("")

#keyboard
rows = ["QWERTYUIOP",
        "ASDFGHJKL",
        "ZXCVBNM"]

#Dictionary
keyboard= {letter: ' ' for row in rows for letter in row}

#colour code ANSII
green = '\033[92m'
yellow = '\033[93m'
gray = '\033[90m'
reset = '\033[0m'

#game
attempt = 0


while attempt < 6:
    # Display the keyboard before each guess
    for row in rows:
        row_display = ""
        for letter in row:
            if keyboard[letter] == 'G':
                row_display += green + letter + reset + "  "  #Green
            elif keyboard[letter] == 'Y':
                row_display += yellow + letter + reset + "  "  #Yellow
            elif keyboard[letter] == 'X':
                row_display += gray + letter + reset + "  "  #Gray
            else:
                row_display += letter + "  "  #Unchanged
        print(row_display)
    print("\n")

    #input

    guess = input("Attempt " + str(attempt + 1) + ": ").lower()

    if len(guess) != 5 or not guess.isalpha() :
        print("Please enter a 5-letter word.")
        continue

    result = ""
    correct_letters = []
    misplaced_letters = []

    #guess
    for i in range(5):
        if guess[i] == w[i]:
            result += green + guess[i] + reset  #Green
            correct_letters.append(guess[i])
            keyboard[guess[i].upper()] = 'G'  #Making keyboard letter to green
        elif guess[i] in w:
            result += yellow + guess[i] + reset  #Yellow
            misplaced_letters.append(guess[i])
            if keyboard[guess[i].upper()] != 'G':  # Don't overwrite green letters
                keyboard[guess[i].upper()] = 'Y'  # Making keyboard letter to yellow
        else:
            result += gray + guess[i] + reset  #Gray
            keyboard[guess[i].upper()] = 'X'  # Making keyboard letter to gray

    print(result)

    if guess == w:
        print("Congratulations! " + w + " was the correct word!")
        break

    attempt += 1

#no correct guess in 6 attempts
if attempt == 6:
    print("Sorry! You've used all your attempts. The word was " + w + ".")
```
